```python
import sys
# import sys.path.append('./')
from BPMQ_model import TISRAW2BPMQ_model, train_TISRAW2BPMQ_model
from TIS161_coeffs import TIS161_coeffs
```

```python
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
import torch
_dtype = torch.float32

import warnings
warnings.filterwarnings("ignore", category=UserWarning, message="To copy construct from a tensor")
```

```python
BPMs = ['BDS_BTS:BPM_D5513','BDS_BTS:BPM_D5565']
```

```python
data_path = './'
training_data = pickle.load(open(data_path+'BDS_beamQ_training_data_from_ISAAC_PMver0.pkl','rb'))
```

```python
for k,v in training_data.items():
    print(f'{k}:   \t type={type(v).__name__}, \t lenth={len(v)}')
```

```python
lrelpath = []
lmeas  = []
lsim   = []
lfmeas = []
lfsim  = []

for relpath, meas, sim in zip(training_data['rel_path'],training_data['measurement'],training_data['simulation']):
    if meas['BDS_BTS:BPM_D5513']['Q_err'].mean() > 1 or meas['BDS_BTS:BPM_D5565']['Q_err'].mean() > 1 :
        continue
    lrelpath.append(relpath)
    mask1 = meas.notna().all(axis=1)
    mask2 = sim.notna().all(axis=1)
    # mask3 = (meas[[(bpm,'beamQ') for bpm in BPMs]].abs() < 30).all(axis=1)
    # mask4 = ( sim[[(bpm,'beamQ') for bpm in BPMs]].abs() < 30).all(axis=1)
    combined_mask = mask1 & mask2 #& mask3 & mask4
    filtered_meas, filtered_sim = meas[combined_mask], sim[combined_mask]
    lmeas.append(meas)
    lsim.append(sim)
    lfmeas.append(filtered_meas)
    lfsim.append(filtered_sim)

# meas = pd.concat(lmeas,axis=0,ignore_index=True)
# sim  = pd.concat(lsim,axis=0,ignore_index=True) 
filtered_meas = pd.concat(lfmeas,axis=0,ignore_index=True)
filtered_sim  = pd.concat(lfsim,axis=0,ignore_index=True)
```

# normalize TISRAW

```python
for bpm in BPMs:
    # Scale and normalize the data for all TISRAW{i} columns 
    new_values =  filtered_meas[bpm][[f"TISRAW{i}" for i in range(1, 5)]] / filtered_meas[bpm]["MAG"].values[:, None] * TIS161_coeffs[bpm][None,:] *1.1e-11 + 0.5
    filtered_meas[[(bpm, f"TISRAW{i}") for i in range(1, 5)]] = new_values
```

```python
fig,ax = plt.subplots(1,2,figsize=(8, 2))
for ibpm,bpm in enumerate(BPMs):
    for i in range(1, 5):
        min_values = [min(lst) for lst in filtered_meas[bpm][f'TISRAW{i}']]
        ax[ibpm].hist(min_values, bins=50, alpha=0.6, label=f"TISRAW{i}")
    ax[ibpm].set_title(f"MinNormTISRAW {bpm[8:]}")
    ax[ibpm].set_ylabel("Count")
    ax[ibpm].legend()
```

# discard outlier

```python
masks = {}
fig,ax = plt.subplots(1,2,figsize=(8,2))
n_data_ = len(filtered_sim[bpm])
n_data  = int(0.98*n_data_)
for i,bpm in enumerate(BPMs):
    loss =  np.sqrt((filtered_meas[bpm]['beamQ']-filtered_sim[bpm]['beamQ'])**2/(np.abs(filtered_sim[bpm]['beamQ'])+4)**2)
    cut = np.sort(loss)[n_data] 
    ax[0].plot(filtered_sim[bpm]['beamQ'],loss,'.',label=bpm[8:])
    ax[0].plot([-30,30],[cut,cut],c='C'+str(i),ls='--')
    ax[1].hist(loss,label=bpm[8:],bins=32)
    ax[1].plot([cut,cut],[0,50],c='C'+str(i),ls='--')
    masks[bpm] = loss < cut

ax[0].set_xlabel('simulated $\sigma_x^2-\sigma_y^2 (mm^2)$')
ax[0].set_ylabel('norm. BPMQerror')
ax[0].legend()
ax[1].legend()
ax[1].set_yscale('log')
```

```python
filtered_meas_backup = filtered_meas
filtered_sim_backup  = filtered_sim
filtered_meas = {}
filtered_sim  = {}
for bpm in BPMs:
    filtered_meas[bpm] = filtered_meas_backup[bpm][masks[bpm]]
    filtered_sim [bpm] = filtered_sim_backup [bpm][masks[bpm]]
```

```python
all_T1 = {}
all_T2 = {}
all_T3 = {}
all_T4 = {}
all_xy = {}
all_Qsim = {}
all_Qmeas = {}
all_Qerr = {}
np.random.seed(0)

n_data = {}
for bpm in BPMs:
    n_data[bpm] = len(filtered_meas[bpm])
    p = np.random.permutation(np.arange(n_data[bpm]))
    all_T1   [bpm] = torch.tensor(np.stack(filtered_meas[bpm]['TISRAW1'].values[p]),dtype=_dtype)
    all_T2   [bpm] = torch.tensor(np.stack(filtered_meas[bpm]['TISRAW2'].values[p]),dtype=_dtype)
    all_T3   [bpm] = torch.tensor(np.stack(filtered_meas[bpm]['TISRAW3'].values[p]),dtype=_dtype)
    all_T4   [bpm] = torch.tensor(np.stack(filtered_meas[bpm]['TISRAW4'].values[p]),dtype=_dtype)
    all_xy   [bpm] = torch.tensor(np.stack(filtered_meas[bpm][['XPOS','YPOS']].values[p]),dtype=_dtype)
    all_Qsim [bpm] = torch.tensor(filtered_sim [bpm]['beamQ'  ].values[p],dtype=_dtype)
    all_Qmeas[bpm] = torch.tensor(filtered_meas[bpm]['beamQ'  ].values[p],dtype=_dtype)
    all_Qerr [bpm] = torch.tensor(filtered_meas[bpm]['beamQ_err'].values[p],dtype=_dtype)
```

# train model

```python
def find_factors(n):
    factors = []
    for i in range(1, n + 1):
        if n % i == 0:
            factors.append(i)
    return factors
```

```python
from math import ceil

n_train = {} 
n_val = {}
n_batch_train = {}
batch_size = {}

for bpm in BPMs:
    n_train[bpm] = int(0.7*n_data[bpm])
    factors = find_factors(n_train[bpm])
    batch_size[bpm] = factors[ceil(0.3*len(factors))]
    n_batch_train [bpm]= int(n_train[bpm]/batch_size[bpm])
    print(bpm, n_data[bpm], n_train[bpm], batch_size[bpm], n_batch_train[bpm])
```

```python
train_T1 = {}
train_T2 = {}
train_T3 = {}
train_T4 = {}
train_xy = {}
train_Qsim = {}
train_Qmeas = {}
train_Qerr = {}

val_T1 = {}
val_T2 = {}
val_T3 = {}
val_T4 = {}
val_xy = {}
val_Qsim = {}
val_Qmeas = {}
val_Qerr = {}

for bpm in BPMs:
    n_val = n_data[bpm] - n_train[bpm]
    
    val_T1[bpm] = all_T1[bpm][:n_val]
    val_T2[bpm] = all_T2[bpm][:n_val]
    val_T3[bpm] = all_T3[bpm][:n_val]
    val_T4[bpm] = all_T4[bpm][:n_val]
    val_xy[bpm] = all_xy[bpm][:n_val]
    val_Qsim[bpm] = all_Qsim[bpm][:n_val]
    val_Qerr[bpm] = all_Qerr[bpm][:n_val]
    val_Qmeas[bpm] = all_Qmeas[bpm][:n_val]
    
    train_T1[bpm] = all_T1[bpm][n_val:]
    train_T2[bpm] = all_T2[bpm][n_val:]
    train_T3[bpm] = all_T3[bpm][n_val:]
    train_T4[bpm] = all_T4[bpm][n_val:]
    train_xy[bpm] = all_xy[bpm][n_val:]
    train_Qsim[bpm] = all_Qsim[bpm][n_val:]
    train_Qerr[bpm] = all_Qerr[bpm][n_val:]
    train_Qmeas[bpm] = all_Qmeas[bpm][n_val:]
```

```python
models = {}
history = {}

for bpm in BPMs:
    fname = bpm.replace('_D','')[-7:] 
    model_info = {'F':2,
                  'dtype':_dtype}
    model = TISRAW2BPMQ_model(**model_info)
    hist_,_,_ = train_TISRAW2BPMQ_model(
        model,10000,2e-3,
        train_T1[bpm],train_T2[bpm],train_T3[bpm],train_T4[bpm],train_xy[bpm],train_Qsim[bpm],train_Qerr[bpm],
        train_T1[bpm],train_T2[bpm],train_T3[bpm],train_T4[bpm],train_xy[bpm],train_Qsim[bpm],train_Qerr[bpm],
        batch_size = batch_size[bpm],
        fname_model = fname+'/model_TISRAW2BPMQ.pt',
        fname_history = fname+'/history_TISRAW2BPMQ.pkl',
        verbose = True
    )
    models[bpm] = model
    history[bpm] = hist_
```

```python
fig, ax = plt.subplots(1,2,figsize=(10,3),dpi=96)
for i, bpm in enumerate(BPMs):
    ax[i].set_title(bpm)
    ax[i].semilogy(history[bpm]['train_loss'],label='train_loss')
    ax[i].semilogy(history[bpm]['val_loss'],label='val_loss')
    ax[i].set_xlabel('epoch')
    ax[i].legend()

plt.tight_layout()
```

```python
train_Qpred = {}
with torch.no_grad():
    for bpm in BPMs:
        model = models[bpm]
        feature, beamQpred = model(train_T1[bpm],train_T2[bpm],train_T3[bpm],train_T4[bpm],train_xy[bpm])
        train_Qpred[bpm] = beamQpred
        
fig, ax = plt.subplots(2, 2, figsize=(8, 8),dpi=96)
for i, bpm in enumerate(BPMs):

    ax[i,0].scatter(train_Qsim[bpm], train_Qmeas[bpm],label='original')
    ax[i,0].scatter(train_Qsim[bpm], train_Qpred[bpm],label='NN')
    ax[i,0].plot([-25, 25], [-25, 25], color='k')
    ax[i,0].set_xlabel('simulated $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i,0].set_ylabel('predicted $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i,0].legend()

    # Second subplot in the row
    ax[i,1].scatter(train_Qsim[bpm], train_Qmeas[bpm]-train_Qsim[bpm],label='original')
    ax[i,1].scatter(train_Qsim[bpm], train_Qpred[bpm]-train_Qsim[bpm],label='NN')
    ax[i,1].set_xlabel('simulated $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i,1].set_ylabel('prediction error of $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i,1].set_xlim(-25,25)
    ax[i,1].set_ylim(-7, 7)
    ax[i,1].legend()
    
    # Set a single title for each row on the first subplot
    ax[i, 0].set_title(bpm, fontsize=16)

plt.tight_layout()
```

```python
val_Qpred = {}
with torch.no_grad():
    for bpm in BPMs:
        model = models[bpm]
        feature, beamQpred =model(val_T1[bpm],val_T2[bpm],val_T3[bpm],val_T4[bpm],val_xy[bpm])
        val_Qpred[bpm] = beamQpred.view(-1).detach()
        
fig, ax = plt.subplots(2, 2, figsize=(8, 8),dpi=96)
for i, bpm in enumerate(BPMs):

    ax[i,0].scatter(val_Qsim[bpm], val_Qmeas[bpm],label='original')
    ax[i,0].scatter(val_Qsim[bpm], val_Qpred[bpm],label='NN')
    ax[i,0].plot([-25, 25], [-25, 25], color='k')
    ax[i,0].set_xlabel('simulated $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i,0].set_ylabel('predicted $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i,0].legend()

    # Second subplot in the row
    ax[i,1].scatter(val_Qsim[bpm], val_Qmeas[bpm]-val_Qsim[bpm],label='original')
    ax[i,1].scatter(val_Qsim[bpm], val_Qpred[bpm]-val_Qsim[bpm],label='NN')
    ax[i,1].set_xlabel('simulated $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i,1].set_ylabel('prediction error of $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i,1].set_xlim(-25,25)
    ax[i,1].set_ylim(-7, 7)
    ax[i,1].legend()
    
    # Set a single title for each row on the first subplot
    ax[i, 0].set_title(bpm, fontsize=16)

plt.tight_layout()
```

# use NN as feature extractor and use GP for BPMQ pred model with uncertainty

```python
from botorch.models import SingleTaskGP
from botorch.fit import fit_gpytorch_mll
from gpytorch.mlls import ExactMarginalLogLikelihood
```

```python
GPs = {}
for bpm in BPMs:
    fname = bpm.replace('_D','')[-7:] 
    with torch.no_grad():
        model = models[bpm].eval()
        feature, beamQpred = model(train_T1[bpm],train_T2[bpm],train_T3[bpm],train_T4[bpm],train_xy[bpm])
        
        beamQerr = torch.sqrt( 0.5*(train_Qmeas[bpm] - train_Qsim[bpm])**2 +  0.25*train_Qerr[bpm]**2 + 0.25*0.2**2)
        
        x_max, x_min = feature.max(dim=0).values.view(1,-1), feature.min(dim=0).values.view(1,-1)
        y_mean, y_std = beamQpred.mean(), beamQpred.std()
        train_x = (feature - x_min)/(x_max-x_min)
        train_y = (beamQpred - y_mean)/y_std
        train_yvar = (beamQerr/y_std)**2
        train_x = train_x.to(torch.float64)
        train_y = train_y.to(torch.float64).view(-1,1)
        train_yvar = train_yvar.to(torch.float64).view(-1,1)
    gp = SingleTaskGP(train_x, train_y, train_Yvar = train_yvar)
    mll = ExactMarginalLogLikelihood(gp.likelihood, gp)
    fit_gpytorch_mll(mll)
        
    torch.save({
        "state_dict": gp.state_dict(),
        "train_x": train_x, 
        "train_y": train_y,
        "train_yvar": train_yvar,
        "x_max":x_max,
        "x_min":x_min,
        "y_mean":y_mean,
        "y_std":y_std,
        }, fname+'/gp_TISRAW2BPMQ.pt')
```

```python
train_Qpred = {}
val_Q_GPpred = {}
train_Q_GPpred = {}

with torch.no_grad():
    for bpm in BPMs:
        fname = bpm.replace('_D', '')[-7:] 
        model = models[bpm]
        
        # Load the checkpoint
        checkpoints = torch.load(fname + '/gp_TISRAW2BPMQ.pt')
        
        # Extract data from the checkpoint
        train_x = checkpoints['train_x']
        train_y = checkpoints['train_y']
        train_yvar = checkpoints['train_yvar']
        x_max = checkpoints['x_max']
        x_min = checkpoints['x_min']
        y_mean = checkpoints['y_mean']
        y_std = checkpoints['y_std']
        
        # Initialize and load the GP model
        gp = SingleTaskGP(train_x, train_y, train_yvar)
        gp.load_state_dict(checkpoints['state_dict'])
        
        # Get features and predictions from the model
        train_feature, train_beamQpred = model(train_T1[bpm], train_T2[bpm], train_T3[bpm], train_T4[bpm], train_xy[bpm])
        val_feature, val_beamQpred = model(val_T1[bpm], val_T2[bpm], val_T3[bpm], val_T4[bpm], val_xy[bpm])
        
        # Scale the features
        train_x_scaled = (train_feature - x_min) / (x_max - x_min)
        val_x_scaled = (val_feature - x_min) / (x_max - x_min)
        
        # Get GP predictions
        train_ypred = gp.posterior(train_x_scaled)
        val_ypred = gp.posterior(val_x_scaled)
        
        # Unscale the predictions
        train_Q_GPpred[bpm] = {
            'mean': train_ypred.mean * y_std + y_mean,
            'std': train_ypred.stddev * y_std,
        }
        val_Q_GPpred[bpm] = {
            'mean': val_ypred.mean * y_std + y_mean,
            'std': val_ypred.stddev * y_std,
        }
```

##### GP prediction on train data

```python
# Plotting
fig, ax = plt.subplots(2, 3, figsize=(10, 7), dpi=128)
for i, bpm in enumerate(BPMs):
    # Flatten the mean and std arrays
    mean = train_Q_GPpred[bpm]['mean'].flatten()
    std = train_Q_GPpred[bpm]['std'].flatten()   # 2-sigma error bars
    
    # First subplot in the row
    ax[i, 0].scatter(train_Qsim[bpm], train_Qmeas[bpm], label='original')
    ax[i, 0].errorbar(train_Qsim[bpm], mean, std, fmt='.', linestyle='none', capsize=5, capthick=2, label='GP 2$\sigma$', color='C1', alpha=0.5)
    ax[i, 0].plot([-25, 25], [-25, 25], color='k')
    ax[i, 0].set_xlabel('simulated $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i, 0].set_ylabel('predicted $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i, 0].legend()

    # Second subplot in the row
    ax[i, 1].scatter(train_Qsim[bpm], train_Qmeas[bpm] - train_Qsim[bpm], label='original')
    ax[i, 1].errorbar(train_Qsim[bpm], mean - train_Qsim[bpm], std, fmt='.', linestyle='none', capsize=5, capthick=2, label='GP 2$\sigma$', color='C1', alpha=0.5)
    ax[i, 1].set_xlabel('simulated $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i, 1].set_ylabel('prediction error of $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i, 1].set_xlim(-25, 25)
    ax[i, 1].set_ylim(-7, 7)
    ax[i, 1].legend()

    # Third subplot in the row
    err = torch.abs(mean - train_Qsim[bpm])
    unc = 2 * std
    
    ax[i, 2].scatter(err, unc)
    xmin = min(min(ax[i, 2].get_xlim()), min(ax[i, 2].get_ylim()))
    xmax = max(max(ax[i, 2].get_xlim()), max(ax[i, 2].get_ylim()))
    ax[i, 2].set_xlabel('error of mean prediction $(mm^2)$')
    ax[i, 2].set_ylabel('uncertainty ($2\sigma$) of prediction $(mm^2)$')
    ax[i, 2].plot([xmin, xmax], [xmin, xmax], 'k')
    
    # Set a single title for each row on the first subplot
    ax[i, 0].set_title(bpm, fontsize=16)

plt.tight_layout()
plt.show()
```

```python
# Plotting
fig, ax = plt.subplots(2, 3, figsize=(10, 7), dpi=128)
for i, bpm in enumerate(BPMs):
    # Flatten the mean and std arrays
    mean = val_Q_GPpred[bpm]['mean'].flatten()
    std = val_Q_GPpred[bpm]['std'].flatten() * 2  # 2-sigma error bars
    
    # First subplot in the row
    ax[i, 0].scatter(val_Qsim[bpm], val_Qmeas[bpm], label='original')
    ax[i, 0].errorbar(val_Qsim[bpm], mean, std, fmt='.', linestyle='none', capsize=5, capthick=2, label='GP 2$\sigma$', color='C1',alpha=0.5)
    ax[i, 0].plot([-25, 25], [-25, 25], color='k')
    ax[i, 0].set_xlabel('simulated $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i, 0].set_ylabel('predicted $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i, 0].legend()

    # Second subplot in the row
    ax[i, 1].scatter(val_Qsim[bpm], val_Qmeas[bpm] - val_Qsim[bpm], label='original')
    ax[i, 1].errorbar(val_Qsim[bpm], mean - val_Qsim[bpm], std, fmt='.', linestyle='none', capsize=5, capthick=2, label='GP 2$\sigma$', color='C1',alpha=0.5)
    ax[i, 1].set_xlabel('simulated $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i, 1].set_ylabel('prediction error of $\sigma_x^2-\sigma_y^2 (mm^2)$')
    ax[i, 1].set_xlim(-25, 25)
    ax[i, 1].set_ylim(-7, 7)
    ax[i, 1].legend()

    err = torch.abs(mean - val_Qsim[bpm])
    unc = 2*std
    
    ax[i, 2].scatter(err,unc)
    xmin = min( min(ax[i,2].get_xlim()), min(ax[i,2].get_ylim()) )
    xmax = max( max(ax[i,2].get_xlim()), max(ax[i,2].get_ylim()) )
    ax[i, 2].set_xlabel('error of mean prediction $(mm^2)$')
    ax[i, 2].set_ylabel('uncertainty ($2\sigma$) of prediction $(mm^2)$')
    ax[i, 2].plot([xmin,xmax],[xmin,xmax],'k')
    
    
    # Set a single title for each row on the first subplot
    ax[i, 0].set_title(bpm, fontsize=16)

plt.tight_layout()
plt.show()
```
