```python
import sys
sys.path.append('../BPMQ')

from LinearControl import LinearModel, LinearControl, LinearControl_virtual_evaluator
import numpy as np
```

```python
input_CSETs = ["BDS_BTS:PSC2_D5496", "BDS_BTS:PSC1_D5496"]
output_RDs = ["BDS_BTS:BPM_D5513:XPOS_RD", "BDS_BTS:BPM_D5513:YPOS_RD"]
input_dim = len(input_CSETs)
output_dim = len(output_RDs)
```
